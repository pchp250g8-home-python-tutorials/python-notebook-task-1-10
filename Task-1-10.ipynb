При каком натуральном значении n дробь (2 * n ** 3 - 2 * n ** 2 - 11 * n + 3) / (n + 1) является натуральным чилом? 
В ответ записать сумму этих значний n.

```python

# -- способ № 1 --
from sympy import pdiv
from sympy.abc import n
from sympy import divisors
from sympy import apart
from numpy import sum
from sympy import solve
from sympy import Eq
from sympy import numer
from sympy import denom

expr1 = (2 * n ** 3 - 2 * n ** 2 - 11 * n + 3) / (n + 1)
expr2 = apart(expr1, n)
n_len = len(expr2.args)
expr3 = expr2.args[n_len - 1]
expr4 = numer(expr3)
print("Initial fract:",expr1)
print("Fract with integer part and with regular fract:", expr2)
print("Regular fract:",expr3)
divs = divisors(expr4)
print("divisors of the regular fract:", divs)
c = 0; s = 0
for d in divs:
    eq1 = Eq(expr3, d)
    s_eq1 = str(eq1.lhs) + " = " + str(eq1.rhs)
    sol = solve(eq1,n)
    print("Equation N",(c+1),":",s_eq1,"Solution:",sol)
    m = sol.pop()
    if(m > 0):
        s += m
    c += 1
print("The sum of numbers is:",s)
```
